```python
import numpy as np
import matplotlib.pyplot as plt
import sklearn
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer, TfidfVectorizer
from sklearn import svm
from sklearn.model_selection import train_test_split, cross_val_score, KFold, ShuffleSplit, GridSearchCV, RandomizedSearchCV
import nltk
from nltk import word_tokenize
import pandas as pd
```

# Getting a corpus

```python
dataset = fetch_20newsgroups(shuffle=True, random_state=1,
                             remove=('headers', 'footers', 'quotes'))
```

# Classifying topics

## Obtain labeled examples

```python
num_of_features = 5000
num_of_examples = 10000
```

```python
count_vectorizer = CountVectorizer(stop_words='english', max_df=0.95, min_df=2, max_features=num_of_features)
```

```python
corpus = dataset.data[:num_of_examples]
```

```python
print(corpus[100])
```

```python
bows = count_vectorizer.fit_transform(corpus)
bows
```

```python
vocabulary = count_vectorizer.get_feature_names_out()
print(vocabulary[100])
```

```python
print(bows[100])
```

```python
targets = dataset.target[:num_of_examples]
```

## Produce training and test data

```python
#targets = targets.reshape((-1, 1))
#targets.shape
```

```python
from tensorflow.keras.utils import to_categorical

y = to_categorical(targets,num_classes=len(set(targets)))
print(targets[:3])
print('  |\n',' V')
print(y[:3])
print('Shape:', y.shape)
```

```python
X_train, X_test, y_train, y_test = train_test_split(bows, y, test_size=0.1)
```

```python
X_train[1], y_train[1]
```

```python
X_train = X_train.toarray()
y_train = y_train
X_test = X_test.toarray()
y_test = y_test
```

```python
X_train
```

## Classify and look at results

```python
import tensorflow as tf
model = tf.keras.models.Sequential([
    tf.keras.layers.Dense(20, activation=tf.keras.activations.relu),
    tf.keras.layers.Dense(20, activation=tf.keras.activations.softmax)
])
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
model.fit(X_train, y_train, epochs=10, validation_split=0.1)
```

```python
model.evaluate(X_test, y_test)
```
